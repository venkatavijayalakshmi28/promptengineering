```python
from google.colab import userdata

try:
    GROQ_API_KEY = userdata.get("GROQ_API_KEY")

    if not GROQ_API_KEY:
        raise ValueError("GROQ_API_KEY not found.")

    print("✅ API Key Loaded Successfully.")

except Exception as e:
    print(f"❌ Error: {e}")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq
```

```python
prompt = ChatPromptTemplate.from_template(
    "Write a very short story (2-3 sentences) about {topic}."
)
```

```python
llm = ChatGroq(
    groq_api_key=GROQ_API_KEY,
    model_name="llama-3.1-8b-instant",
    temperature=0.5
)
```

```python
parser = StrOutputParser()
```

```python
chain = prompt | llm | parser
```

```python
topics = [
    "a dragon and a knight",
    "a lost spaceship"
]
```

```python
for topic in topics:

    print("="*80)
    print("Topic:")
    print(topic)

    story = chain.invoke(
        {
            "topic": topic
        }
    )

    print("\nStory:")
    print(story)
    print()
```

```python
print("="*80)
print("ANALYSIS")
print("="*80)

print("""
1. A ChatPromptTemplate inserts the topic into the prompt.

2. The ChatGroq LLM generates a short story.

3. StrOutputParser converts the response into plain text.

4. The LCEL pipe syntax (prompt | llm | parser)
creates a simple and readable processing chain.

5. The same chain was reused for multiple topics.
""")
```
