```python
from google.colab import userdata
from openai import OpenAI

try:
    api_key = userdata.get("GROQ_API_KEY")

    if not api_key:
        raise ValueError("GROQ_API_KEY not found in Colab Secrets.")

    client = OpenAI(
        api_key=api_key,
        base_url="https://api.groq.com/openai/v1"
    )

    print("✅ Connected to Groq successfully!")

except Exception as e:
    print(f"❌ Error: {e}")
```

```python
MODEL = "llama-3.3-70b-versatile"
TEMPERATURE = 0.5
MAX_TOKENS = 200

def ask_llm(prompt):
    try:
        response = client.chat.completions.create(
            model=MODEL,
            temperature=TEMPERATURE,
            max_tokens=MAX_TOKENS,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        return response.choices[0].message.content

    except Exception as e:
        return f"❌ Error: {e}"
```

```python
# Iteration 0 - Baseline
prompt0 = (
    "Explain photosynthesis to a 10-year-old."
)

# Iteration 1 - Add role, analogy and brevity
prompt1 = (
    "You are a friendly science teacher. "
    "Explain photosynthesis to a 10-year-old using a simple kitchen analogy. "
    "Keep the explanation to exactly 3 short sentences and avoid technical jargon."
)

# Iteration 2 - Add mnemonic and diagram description
prompt2 = (
    "You are a friendly science teacher. "
    "Explain photosynthesis to a 10-year-old using a simple kitchen analogy. "
    "Keep the explanation to exactly 3 short sentences, avoid technical jargon, "
    "include a memorable mnemonic, and describe a simple text-only diagram."
)

prompts = [prompt0, prompt1, prompt2]
responses = []
```

```python
for i, prompt in enumerate(prompts):

    print("="*70)
    print(f"ITERATION {i}")
    print("="*70)

    print("\nPrompt:")
    print(prompt)

    response = ask_llm(prompt)

    responses.append(response)

    print("\nResponse:")
    print(response)

    print("\nAnalysis:")

    if i == 0:
        print("- Basic explanation.")
        print("- May be lengthy or include technical words.")
        print("- No analogy or formatting instructions.")

    elif i == 1:
        print("- Added role assignment.")
        print("- Added analogy.")
        print("- Limited response length.")
        print("- Easier for children to understand.")

    else:
        print("- Added mnemonic for easier memory.")
        print("- Added text-only diagram description.")
        print("- Better engagement and visualization.")

    print("\n")
```

```python
print("="*70)
print("FINAL SUMMARY")
print("="*70)

print("\nIteration 0:")
print("- Simple prompt with no guidance.")
print("- Response depends entirely on the model.")

print("\nIteration 1:")
print("- Improved clarity using a teacher role.")
print("- Added analogy and sentence limit.")
print("- Better suited for a 10-year-old.")

print("\nIteration 2:")
print("- Added mnemonic and diagram description.")
print("- Improved memorability and visualization.")
print("- Most effective version.")

print("\nOverall Conclusion:")
print("Prompt engineering significantly improves response quality.")
print("Adding role, constraints, analogies, and formatting instructions")
print("produces responses that are clearer, more engaging, and better")
print("matched to the target audience.")
```
