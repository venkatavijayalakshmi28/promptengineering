```python
from google.colab import userdata
from groq import Groq

try:
    api_key = userdata.get("GROQ_API_KEY")

    if not api_key:
        raise ValueError("GROQ_API_KEY not found.")

    client = Groq(api_key=api_key)

    print("✅ Connected Successfully!")

except Exception as e:
    print(f"❌ Error: {e}")
```

```python
MODEL = "llama-3.3-70b-versatile"
TEMPERATURE = 0.2
MAX_TOKENS = 250

def ask_llm(prompt):
    try:
        response = client.chat.completions.create(
            model=MODEL,
            temperature=TEMPERATURE,
            max_tokens=MAX_TOKENS,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        return response.choices[0].message.content

    except Exception as e:
        return f"API Error: {e}"
```

```python
unstructured_prompt = """
Describe Python, Java, and C++ in a short paragraph.
"""

structured_prompt = """
Create a markdown table with the following columns:

Language
Typing (Static/Dynamic)
Primary Use Case
Difficulty (Beginner/Intermediate/Advanced)

Include exactly these three programming languages:

Python
Java
C++

Do not include any explanation outside the table.
"""
```

```python
unstructured_response = ask_llm(unstructured_prompt)

structured_response = ask_llm(structured_prompt)
```

```python
print("="*80)
print("UNSTRUCTURED OUTPUT")
print("="*80)

print("\nPrompt:")
print(unstructured_prompt)

print("\nResponse:")
print(unstructured_response)

print("\n\n")

print("="*80)
print("STRUCTURED OUTPUT (MARKDOWN TABLE)")
print("="*80)

print("\nPrompt:")
print(structured_prompt)

print("\nResponse:")
print(structured_response)
```

```python
print("="*80)
print("ANALYSIS")
print("="*80)

print("\n1. Readability")
print("- The unstructured response presents the information as a paragraph.")
print("- The structured response organizes information into a markdown table.")

print("\n2. Comparison")
print("- The table makes it easier to compare typing, use case, and difficulty.")
print("- Each programming language is clearly separated into its own row.")

print("\n3. Structure")
print("- The markdown table follows the required columns.")
print("- The response is easier to scan than a paragraph.")

print("\nSummary")
print("Structured prompting improves readability and comparison by organizing information into a clear markdown table while preserving the same content.")
```
