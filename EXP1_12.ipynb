```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
sample_text = """
Climate change refers to long-term changes in Earth's temperature and weather patterns.
Human activities, especially burning fossil fuels such as coal, oil, and natural gas,
release greenhouse gases into the atmosphere.

These greenhouse gases trap heat and increase global temperatures.
As a result, glaciers melt, sea levels rise, and extreme weather events such as floods,
storms, droughts, and heatwaves become more frequent.

Climate change also affects biodiversity, agriculture, water resources,
and human health. Reducing greenhouse gas emissions,
using renewable energy, protecting forests, and improving energy efficiency
are important ways to reduce climate change.

Many countries are working together through international agreements
to slow global warming and protect future generations.
"""
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=30
)

documents = [Document(page_content=sample_text)]

chunks = text_splitter.split_documents(documents)

print("Number of chunks:", len(chunks))
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embeddings model loaded successfully.")
```

```python
vectorstore = FAISS.from_documents(
    chunks,
    embeddings
)

print("FAISS Vector Store Created Successfully.")
```

```python
query = "What are the effects of climate change?"

results = vectorstore.similarity_search_with_score(
    query,
    k=2
)
```

```python
print("="*80)
print("QUERY")
print("="*80)

print(query)

print("\n")

print("="*80)
print("TOP 2 RETRIEVED CHUNKS")
print("="*80)

for i, (doc, score) in enumerate(results, start=1):

    print(f"\nResult {i}")
    print("-"*60)

    print("Similarity Score:", score)

    print("\nRetrieved Chunk:\n")
    print(doc.page_content)
```

```python
print("="*80)
print("ANALYSIS")
print("="*80)

print("Number of Chunks Created:", len(chunks))

print("\nThe vector store successfully indexed the document.")

print("\nThe similarity search retrieved the two most relevant chunks for the query.")

print("\nSmaller distance scores indicate higher similarity in FAISS.")

print("\nThis demonstrates the indexing and retrieval stage of a Retrieval-Augmented Generation (RAG) pipeline.")
```
