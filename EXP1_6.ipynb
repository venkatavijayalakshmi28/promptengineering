```python
from google.colab import userdata
from groq import Groq

try:
    api_key = userdata.get("GROQ_API_KEY")

    if not api_key:
        raise ValueError("GROQ_API_KEY not found.")

    client = Groq(api_key=api_key)

    print("✅ Connected Successfully!")

except Exception as e:
    print(f"❌ Error: {e}")
```

```python
MODEL = "llama-3.3-70b-versatile"
TEMPERATURE = 0.5
MAX_TOKENS = 200

def ask_llm(prompt):
    try:
        response = client.chat.completions.create(
            model=MODEL,
            temperature=TEMPERATURE,
            max_tokens=MAX_TOKENS,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        return response.choices[0].message.content

    except Exception as e:
        return f"API Error: {e}"
```

```python
baseline_prompt = (
    "Explain the importance of exercise."
)

role_prompt = (
    "You are a motivational fitness coach.\n"
    "Explain the importance of exercise."
)

role_negative_prompt = (
    "You are a motivational fitness coach.\n"
    "Do not use any medical jargon.\n"
    "Do not mention weight loss.\n\n"
    "Explain the importance of exercise."
)
```

```python
baseline_response = ask_llm(baseline_prompt)

role_response = ask_llm(role_prompt)

role_negative_response = ask_llm(role_negative_prompt)
```

```python
print("="*70)
print("BASELINE")
print("="*70)

print("\nPrompt:")
print(baseline_prompt)

print("\nResponse:")
print(baseline_response)

print("\n\n")

print("="*70)
print("ROLE-BASED")
print("="*70)

print("\nPrompt:")
print(role_prompt)

print("\nResponse:")
print(role_response)

print("\n\n")

print("="*70)
print("ROLE + NEGATIVE PROMPTING")
print("="*70)

print("\nPrompt:")
print(role_negative_prompt)

print("\nResponse:")
print(role_negative_response)
```

```python
print("="*70)
print("ANALYSIS")
print("="*70)

print("\n1. Tone")
print("- Baseline: Neutral and informative.")
print("- Role-Based: Motivational and encouraging.")
print("- Role + Negative: Motivational while avoiding medical terminology and weight-loss discussion.")

print("\n2. Vocabulary")
print("- Baseline uses general educational language.")
print("- Role-Based uses energetic and positive words.")
print("- Role + Negative uses simple everyday language.")

print("\n3. Constraint Adherence")
print("- Check whether the model avoided medical jargon.")
print("- Check whether weight loss was not mentioned.")
print("- If both constraints were followed, the negative prompting was successful.")

print("\nSummary")
print("Role-based prompting changes the tone and style of the response.")
print("Negative prompting successfully restricts unwanted content while preserving the main message.")
```
